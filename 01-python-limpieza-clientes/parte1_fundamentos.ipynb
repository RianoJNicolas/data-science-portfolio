**Tarea:**

## Paso 9

Aplicar la limpieza a todos los registros garantiza consistencia antes de lanzar el programa de fidelización.

**Tarea:**

1. Elimina los espacios iniciales y finales de los nombres y reemplaza los guiones bajos por espacios.
3. Convierte las edades a números enteros.
5. Separa los nombres y apellidos en una sublista.
6. Guarda el resultado en una nueva lista llamada `usuarios_limpio`
7. Muéstrala en pantalla.

```python
usuarios = [
    ['32415', ' mike_reed ', 32.0, ['ELECTRÓNICA', 'DEPORTE', 'LIBROS'], [894, 213, 173]],
    ['31980', 'kate morgan', 24.0, ['ROPA', 'LIBROS'], [439, 390]],
    ['32156', ' john doe ', 37.0, ['ELECTRÓNICA', 'HOGAR', 'COMIDA'], [459, 120, 99]],
]

usuarios_limpio = []


# Procesa al primer usuario

usuario_id_1 = usuarios[0][0]
usuario_cat_1 = usuarios[0][3]
usuario_gastos_cat_1 = usuarios[0][4]

# 1. Limpia espacios y reemplaza guión bajo
usuario_nombre_1 = usuarios[0][1].strip().replace("_", " ")
# 2. Convierte edad a entero
usuario_edad_1 = int(usuarios[0][2])
# 3. Separa el nombre
usuario_nombre_1 = usuario_nombre_1.split(" ")
# 4. Agrega el usuario limpio a la lista
usuarios_limpio.append([usuario_id_1, usuario_nombre_1, usuario_edad_1, usuario_cat_1, usuario_gastos_cat_1])

# Procesa al segundo usuario

usuario_id_2 = usuarios[1][0]
usuario_cat_2 = usuarios[1][3]
usuario_gastos_cat_2 = usuarios[1][4]
usuario_nombre_2 = usuarios[1][1].split(" ")
usuario_edad_2 = int(usuarios[1][2])
usuarios_limpio.append([usuario_id_2, usuario_nombre_2, usuario_edad_2, usuario_cat_2, usuario_gastos_cat_2])

# Procesa al tercer usuario

usuario_id_3 = usuarios[2][0]
usuario_cat_3 = usuarios[2][3]
usuario_gastos_cat_3 = usuarios[2][4]
usuario_nombre_3 = usuarios[2][1].strip().split(" ")
usuario_edad_3 = int(usuarios[2][2])
usuarios_limpio.append([usuario_id_3, usuario_nombre_3, usuario_edad_3, usuario_cat_3, usuario_gastos_cat_3])

print(usuarios_limpio)

print("\n")

for u in usuarios_limpio:
    print(u)
```
